```python
  # Path to your JSONL file
conversations = []

# Read JSONL file and structure the data
with open(jsonl_file, "r") as file:
    for line in file:
        session = json.loads(line)
        session_id = session["session_id"]
        conversation_text = ""
        for convo in session["conversation"]:
            # Concatenate input and response pairs
            conversation_text += f"Q: {convo['input']}\nA: {convo['response']}\n"
        conversations.append({"session_id": session_id, "conversation": conversation_text.strip()})

# Convert to DataFrame
df = pd.DataFrame(conversations)
print("Dataset loaded and structured.")
```

```python
train_dataset = load_dataset("csv", data_files="train_conversations.csv")["train"]
test_dataset = load_dataset("csv", data_files="test_conversations.csv")["train"]
```

```python
def tokenize_function(examples):
    # Tokenize conversations
    model_inputs = tokenizer(
        examples["conversation"],
        truncation=True,
        padding="max_length",
        max_length=1024
    )
    # Add labels
    model_inputs["labels"] = model_inputs["input_ids"].copy()
    return model_inputs
```

```python
tokenized_train = train_dataset.map(tokenize_function, batched=True)
tokenized_test = test_dataset.map(tokenize_function, batched=True)
```

```python
# Print the first 3 rows of the dataset
print(tokenized_train[:1])
```

```python
training_args = TrainingArguments(
    output_dir="./fine_tuned_llama",
    evaluation_strategy="epoch",
    learning_rate=2e-4,
    per_device_train_batch_size=1,  # Reduce batch size to 1 for memory efficiency
    num_train_epochs=3,
    save_strategy="epoch",
    logging_dir="./logs",
    save_total_limit=2,
    fp16=True,  # Enable mixed precision for lower memory usage
    gradient_accumulation_steps=8,  # Simulate an effective batch size of 32
)
```

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_test,
    tokenizer=tokenizer,

)
```

```python
# Fine-tune the model
# Start fine-tuning
trainer.train()
```

```python
from langchain.prompts import PromptTemplate
from langchain.llms import HuggingFacePipeline
from langchain.memory import ConversationBufferMemory
from langchain_core.runnables.history import RunnableWithMessageHistory
import uuid  # For generating session IDs

# Initialize prompt template
dynamic_prompt_template = PromptTemplate(
    input_variables=["history", "input", "missing_fields"],
    template="""
You are an AI assistant helping to collect information during an emergency call.

### Context:
Your job is to gather the following details:
- Name
- Phone Number
- Address
- Related Address
- Nature of Emergency

If any of these details are missing or incomplete, kindly ask the user for the missing information.

### Conversation History:
{history}

### User Input:
{input}

### AI Response:
"""
)

# Load your fine-tuned model
from transformers import pipeline

# model_name = "./fine_tuned_llama"  # Path to your fine-tuned model
# pipeline_model = pipeline("text-generation", model=model_name, tokenizer=model_name, device=0)
# llm = HuggingFacePipeline(pipeline=pipeline_model)

# Initialize memory
memory = ConversationBufferMemory(return_messages=True)  # Memory with messages support

# Use RunnableWithMessageHistory
conversation_chain = RunnableWithMessageHistory(
    runnable=llm,
    memory=memory,
    prompt=dynamic_prompt_template,
    get_session_history=lambda config: memory.chat_memory  # Correctly provide the memory object
)

# Generate a unique session ID
session_id = str(uuid.uuid4())  # Unique ID for the current session

# Conversation loop
conversation_history = ""
missing_fields = ["name", "phone_number", "address", "related_address", "emergency"]

while True:
    user_input = input("You: ")
    if user_input.lower() in ["exit", "quit"]:
        print("Exiting conversation.")
        break

    # Update missing fields
    missing_fields_str = ", ".join(missing_fields)

    # Invoke the conversation chain
    response = conversation_chain.invoke(
        {
            "history": conversation_history,
            "input": user_input,
            "missing_fields": missing_fields_str,
        },
        {"configurable": {"message_history": memory.chat_memory, "session_id": session_id}}  # Correct memory reference
    )

    print(f"AI: {response}")
    conversation_history += f"User: {user_input}\nAI: {response}\n"
```

```python
# Define required fields
required_fields = ["name", "phone_number", "address", "related_address", "emergency"]
collected_data = {}

def update_missing_fields(collected_data):
    """Returns a list of missing fields."""
    return [field for field in required_fields if field not in collected_data or not collected_data[field]]
```

```python
conversation_history = ""
collected_data = {}

while True:
    # Update missing fields
    missing_fields = update_missing_fields(collected_data)

    # Stop if all fields are collected
    if not missing_fields:
        print("\nAll required information collected:")
        print(collected_data)
        break

    # Get user input and pass it to the model for dynamic questioning
    user_input = input("You: ")
    conversation_history += f"User: {user_input}\n"

    # Pass context to the LangChain ConversationChain
    response = conversation_chain.run({
        "history": conversation_history,
        "input": user_input,
        "missing_fields": ", ".join(missing_fields)
    })

    print(f"AI: {response}")

    # Parse the AI's response to extract field values
    for field in required_fields:
        if field not in collected_data and field in response.lower():
            collected_data[field] = response  # Simplify by directly assigning; parsing can be added here

    conversation_history += f"AI: {response}\n"
```

```python

# Initial conversation context
conversation_context = "Q: How may I help you?\nA:"
print("Model: How may I help you?")

while True:
    # Get user input
    user_input = input("You: ")
    if user_input.lower() in ["exit", "quit"]:
        print("Ending conversation.")
        break

    # Append user's answer to the conversation context
    conversation_context += f" {user_input}\nQ:"

    # Tokenize the updated context
    inputs = tokenizer(
        conversation_context,
        return_tensors="pt",
        truncation=True,
        padding="max_length",
        max_length=512
    ).to("cuda")

    # Generate the model's next question
    with torch.no_grad():
        outputs = model.generate(
            input_ids=inputs["input_ids"],
            attention_mask=inputs["attention_mask"],
            max_new_tokens=50,
            temperature=0.7,
            top_k=50,
            top_p=0.9,
            repetition_penalty=2.0,
            no_repeat_ngram_size=3,
            eos_token_id=tokenizer.eos_token_id,
        )

    # Decode the generated output
    next_question = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # Extract only the new question to avoid repeating the context
    next_question = next_question.replace(conversation_context, "").strip()

    # Print the model's question
    print(f"Model: {next_question}")

    # Update the context with the model's question
    conversation_context += f" {next_question}\nA:"
```
